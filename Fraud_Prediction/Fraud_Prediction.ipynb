# Neural Network Design for Fraud Classification

```python
import matplotlib.pyplot as plt
import networkx as nx
```

Specify the neural network layers

```python
layers = {
    "Input Layer": ["TransactionAmount", "TransactionTime", "MerchantCategory",
                    "CustomerAge", "AccountBalance", "NumberOfTransactionsToday"],
    "Hidden Layer 1": ["H1_Node1", "H1_Node2", "H1_Node3", "H1_Node4"],
    "Hidden Layer 2": ["H2_Node1", "H2_Node2", "H2_Node3"],
    "Output Layer": ["Fraud (0=Genuine, 1=Fraud)"]
}
```

Build the directed network graph

```python
G = nx.DiGraph()
```

Place nodes across each layer

```python
pos = {}
layer_gap = 5
node_gap = 1.5

for i, (layer_name, nodes) in enumerate(layers.items()):
    for j, node in enumerate(nodes):
        G.add_node(node, layer=layer_name)
        pos[node] = (i * layer_gap, j * node_gap)
```

Connect neurons between adjacent layers

```python
layer_names = list(layers.keys())
for i in range(len(layer_names) - 1):
    for src in layers[layer_names[i]]:
        for dst in layers[layer_names[i+1]]:
            G.add_edge(src, dst)
```

Render the neural network layout

```python
plt.figure(figsize=(12,8))
nx.draw(G, pos, with_labels=True, node_size=2500, node_color="skyblue",
        font_size=9, font_weight="bold", arrows=False)
plt.title("Neural Network Architecture for Fraud Detection", fontsize=14)
plt.show()
```
